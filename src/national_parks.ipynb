```python
# Toolkit: run this cell. Don't change it.
import re
import requests
from bs4 import BeautifulSoup

URL = "https://en.wikipedia.org/wiki/List_of_national_parks_of_the_United_States"
PAGE = "parks.html"


def fetch_and_save(url, path):
    # Wikipedia asks scrapers to say who they are. This is an honest name, not a disguise.
    headers = {"User-Agent": "Gann DAwP student scraper (school exercise)"}
    resp = requests.get(url, headers=headers, timeout=20)
    resp.encoding = "utf-8"
    with open(path, "w", encoding="utf-8") as f:
        f.write(resp.text)
    return resp.status_code


def load_soup(path):
    with open(path, encoding="utf-8") as f:
        return BeautifulSoup(f.read(), "html.parser")


def all_numbers(text):
    return [float(n.strip(".").replace(",", "")) for n in re.findall(r"\d[\d,.]*", text)]


def first_number(text):
    found = all_numbers(text)
    return found[0] if found else None
```

```python
def clean_area(text):
    just_number = re.search(r"(.*)\sacres", text).group(1)
    just_number = re.sub(r",","", just_number)
    return float(just_number)
```

```python
def clean_visitors(text):
    return int(re.sub(r",", "", text))
```

```python
def just_year(text):
    return int(re.search(r"\d{4}", text).group(0))
```

```python
def clean_name(text):
    # return text
    return re.sub(r"\s\W$", "", text)
```

```python
def clean_state(text):
    return re.search(r"(\D+)\s\d", text).group(1)
```

```python
parks = {}
for row in rows:
    name = row.find("th").get_text()
    
    tds = row.find_all("td")
    area = tds[3].get_text()
    visitor = tds[4].get_text()
    date = tds[2].get_text()
    state = tds[1].get_text()
    parks[clean_name(name)] = {"year": just_year(date), "area": clean_area(area), "visitors": clean_visitors(visitor), "state" : clean_state(state).strip()}

for key, value in parks.items():
    print(key)

    
```
